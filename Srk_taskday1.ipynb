```python
# Mini project 2 Discount order summary

# store product details, calculate the amount before and after discount and print a clean summary

product_name = "Ubikey"
product_price = 4000
unit = 5
discount_rate = 0.35

amount_before_discount = product_price * unit
discount_amount = amount_before_discount * discount_rate
amount_after_discount = amount_before_discount - discount_amount

print("----summary----")
print("Product Name: " + product_name)
print("Product_price: " + str(product_price))
print("unit: " + str(unit))
print("amount_before_discount: " + str(amount_before_discount))
print("discount_amount: " + str(discount_amount))
print("amount_after_discount: " + str(amount_after_discount))
```
