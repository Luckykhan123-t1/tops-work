## Day 1 — Python Fundamentals

### Theory

**1. What makes Python useful for data-related work?**

Python has simple syntax and a large collection of libraries. Tools such as NumPy, Pandas and Matplotlib make numerical work, data handling and visualization easier.

**2. How are Python 2 and Python 3 different?**

Python 3 is the current Python series and introduced several language improvements. Python 2 is an old, unsupported version and should not be used for new projects.

**3. What does an interpreted language mean?**

An interpreted language is executed through an interpreter while the program is running. Python programs are normally executed by a Python interpreter.

**4. Why is Python commonly used in machine learning?**

Python combines readable syntax with libraries such as scikit-learn, PyTorch and TensorFlow, allowing developers to build and experiment with ML models efficiently.

**5. What is Jupyter Notebook?**

Jupyter Notebook is an interactive environment where code, explanations, outputs and visualizations can be kept together in separate cells.

**6. Compare a .py file with an .ipynb file.**

A .py file normally contains Python source code, while an .ipynb file stores notebook cells containing code, Markdown, outputs and notebook metadata.

**7. What is a Python library?**

A library is reusable code written to solve common problems. For example, NumPy provides numerical tools and Pandas provides data-analysis structures.

**8. What is Anaconda used for?**

Anaconda is a Python distribution commonly used for data science. It bundles Python and many useful packages and development tools.

**9. What is pip?**

pip is a package-management tool used to install and manage Python packages.

**10. What is an IDE?**

An IDE is software that provides tools for writing, running and debugging programs. Examples include VS Code and PyCharm.

**11. What does the Python interpreter do?**

The interpreter reads Python instructions and executes them so that the program can produce a result.

**12. What is interactive programming?**

It means running small pieces of code and immediately checking their results. Notebook cells are a common example.

**13. Why can Python be slower than C for some operations?**

Python has runtime overhead such as dynamic type handling. However, many scientific libraries use optimized lower-level code internally, which makes numerical workloads practical.

**14. What is indentation in Python?**

Indentation is the leading whitespace used to show which statements belong to a block such as an if statement or loop.

# Day 2 — Variables, Types and Code Style

### Theory

**1. What is indentation used for?**

Indentation tells Python which statements belong to the same code block.

**2. Why does Python use indentation for blocks?**

It makes the structure of the program visible and keeps code blocks consistent without requiring curly braces.

**3. Explain dynamic typing.**

Python determines the type of a value at runtime, so a variable does not need a type declaration before assignment.

**4. What is a variable?**

A variable is a name that refers to a value stored by a program.

**5. Are Python variable names case-sensitive?**

Yes. For example, total and Total are different names.

**6. What are keywords?**

Keywords are reserved words with special meaning in Python, such as if, for, while, class and return.

**7. Why cannot keywords normally be used as variable names?**

They already have a defined role in Python's syntax, so using them as ordinary identifiers would conflict with that role.

**8. What is the difference between single-line and multi-line comments?**

A single-line comment begins with #. Several comment lines can be written separately with #; triple-quoted strings are also sometimes used as documentation text.

**9. What is PEP 8?**

PEP 8 is a widely used Python style guide covering readability, spacing, naming and formatting conventions.

**10. Why are naming conventions useful?**

Clear and consistent names make programs easier to read, maintain and understand when several people work on them.

**11. What can happen when indentation is wrong?**

Python can raise an IndentationError, or the program may have a different block structure than intended.

**12. Is Python dynamically and strongly typed?**

Yes. Python is dynamically typed because types are determined at runtime, and strongly typed because incompatible types are not silently mixed in many operations.

**13. Why are comments useful in data projects?**

Comments can document assumptions, transformations and important decisions, helping another developer understand the workflow.

**14. Can a variable refer to values of different types at different times?**

Yes. For example, a name can first refer to an integer and later be assigned a string.

## Mini Project — Discounted Order Summary

Store product details, calculate the amount before discount and after discount, and print a clean summary.

```python
product = "Keyboard"
unit_price = 1800
units = 3
discount_rate = 0.15

subtotal = unit_price * units
discount_amount = subtotal * discount_rate
final_amount = subtotal - discount_amount

print("----- ORDER SUMMARY -----")
print(f"Product: {product}")
print(f"Unit price: ₹{unit_price}")
print(f"Units: {units}")
print(f"Subtotal: ₹{subtotal:.2f}")
print(f"Discount: ₹{discount_amount:.2f}")
print(f"Final amount: ₹{final_amount:.2f}")
```
